```python
from onekey_algo import init_CN
init_CN()
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
import matplotlib.ticker as ticker
from functools import reduce


t1 = pd.read_csv(r'D:\20241225-HuangHaoZhe\Radiology\post/habitat/ch.csv')
t1['group'] = 'Before'
t2 = pd.read_csv(r'D:\20241225-HuangHaoZhe\Radiology\pre\habitat/ch.csv')
t2['group'] = 'After'
a = pd.concat([t1, t2], axis=0)
# a = t1
a.columns=['Nclusters', 'Calinski-Harabasz Score', '_', '__', 'group']
a = a[a['Nclusters'] > 2]
a = a.reset_index(drop=True)
a.to_csv('data/ch.csv', index=False)
fig, ax = plt.subplots(figsize=(14, 9))
# ax.set_ylim(2e5, 8e5)
sns.lineplot(x=a.Nclusters, y=a[a['group'] == 'Before']['Calinski-Harabasz Score'], label='Before', ax=ax, color='r')
ax.legend(loc=0)
ax.set_ylabel('Before Calinski-Harabasz Score')
ax2 = ax.twinx()
# ax2.set_ylim(1e5, 4e5)
sns.lineplot(x=a.Nclusters, y=a[a['group'] == 'After']['Calinski-Harabasz Score'], label='After', ax=ax2, color='g')
ax2.set_ylabel('After Calinski-Harabasz Score')
ax2.legend(loc=2)
# ax = sns.relplot(
#     data=a,
#     x="N-clusters", y="Calinski-Harabasz Score" , kind="line", linewidth=2, color='black', hue='group',
#     palette="ch:r=-.5,l=.75"
# )
ax.figure.set_size_inches(12,6)
plt.savefig('img/Calinski-Harabasz.svg')
```
