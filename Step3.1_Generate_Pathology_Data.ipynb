```python
import os
from glob import glob
import pandas as pd
import cv2
import numpy as np
import os
from onekey_algo import get_param_in_cwd
    
root = get_param_in_cwd('path_dir')
files = {}

samples = pd.DataFrame(glob(os.path.join(root, 'patches', '*', '*.jpg')), columns=['fpath'])
samples['ID'] = samples['fpath'].map(lambda x: os.path.basename(os.path.dirname(x)) + '.nii.gz')
label_data = pd.read_csv(get_param_in_cwd('label_file'))
samples
```

```python
from onekey_algo.custom.utils import print_join_info

print_join_info(samples, label_data)
```

```python
samples = pd.merge(samples, label_data, on='ID', how='inner')
print(len(set(samples['ID'])))
```

```python
train_data = samples[samples['group'] == 'train']
train_data[['fpath', 'label']].to_csv('split_info/ptrain.txt', index=False, header=False, sep='\t')

test_data = samples[samples['group'] != 'train']
test_data[['fpath', 'label']].to_csv('split_info/pval.txt', index=False, header=False, sep='\t')

print(f"Training set: {train_data.shape[0]}({len(np.unique(train_data['ID']))}); test set: {test_data.shape[0]}({len(np.unique(test_data['ID']))})")
```
