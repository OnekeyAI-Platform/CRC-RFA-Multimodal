# Rename files

```python
import os
import pandas as pd
import shutil
from onekey_algo.custom.components.comp2 import split_dataset4sol
from onekey_algo import get_param_in_cwd
from onekey_algo.custom.utils import print_join_info
from onekey_algo.utils import create_img_msk_dir

# model_root: model output directory
root = r'D:\20241225-HuangHaoZhe'
pre_img_dir, pre_msk_dir = create_img_msk_dir(os.path.join(root, 'pre'))
post_img_dir, post_msk_dir = create_img_msk_dir(os.path.join(root, 'post'))
for sample in os.listdir(os.path.join(root, 'data')):
    for mask in os.listdir(os.path.join(root, 'data', sample, 'CT')):
        if os.path.isfile(os.path.join(root, 'data', sample, 'CT', mask)) and not mask.startswith('.'):
            if 'post' in mask.lower():
                shutil.move(os.path.join(root, 'data', sample, 'CT', mask), 
                            os.path.join(post_msk_dir, mask.lower().replace('-post', '')))
            elif 'pre' in mask.lower():
                shutil.move(os.path.join(root, 'data', sample, 'CT', mask), 
                            os.path.join(pre_msk_dir, mask.lower().replace('-pre', '')))
            else:
                print(f"{sample}, {mask}")
```
