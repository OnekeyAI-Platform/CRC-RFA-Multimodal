```python
import pandas as pd
from onekey_algo.custom.utils import key2
from onekey_algo import get_param_in_cwd
import numpy as np
import os

log = pd.read_csv(r'results/Path_ALL_DL_PREDICTIONS.csv')
log.columns = ['group', 'prob', 'pred', 'gt', 'model']
log = log[log['model'] == f'inception_v3_Pathomics']
log['prob'] = log['prob'].round(decimals=2)
os.makedirs('features', exist_ok=True)
print(log.shape)
results = key2.key2histogram(log, group_column='group',histo_columns='prob', norm=True)
results.to_csv(f'features/path_prob_histogram.csv', header=True, index=False)
display(results)

results = key2.key2histogram(log, group_column='group',histo_columns='pred', norm=True)
results.to_csv(f'features/path_pred_histogram.csv', header=True, index=False)
display(results)

results = key2.key2tfidf(log, group_column='group',corpus_columns='prob')
results.to_csv(f'features/path_prob_tfidf.csv', header=True, index=False)
display(results)

results = key2.key2tfidf(log, group_column='group',corpus_columns='pred')
results.to_csv(f'features/path_pred_tfidf.csv', header=True, index=False)
display(results)
```
